```python
df[['Name', 'Molecular_Weight', 'XLogP', 'H-Bond_Donor_Count', 'H-Bond_Acceptor_Count']].isnull().sum()
```

```python
# Apply each rule as a True/False column
df['MW_Pass'] = df['Molecular_Weight'] <=500
df['LogP_Pass'] = df['XLogP'] <=5
df['HBD_Pass'] = df['H-Bond_Donor_Count'] <=5
df['HBA_Pass'] = df['H-Bond_Acceptor_Count'] <=10

# A compound is drug-like if it passes all 4 rules
df['Drug_Like'] = df['MW_Pass'] & df['LogP_Pass'] & df['HBD_Pass'] & df['HBA_Pass']

# See the results
df[['Name', 'Molecular_Weight', 'XLogP', 'H-Bond_Donor_Count', 'H-Bond_Acceptor_Count', 'Drug_Like']]
```

```python
print (df['Drug_Like'].value_counts())
print (f"\n{df['Drug_Like'].sum()} out of {len (df)} compounds are drug-like")
```

```python
plt.figure(figsize=(6,4))
df['Drug_Like'].value_counts().plot(kind='bar', color = ['steelblue', 'salmon'], edgecolor= 'black')
plt.title('Drug-Like Compounds in Withanolides (Lipinski Rule of Five)')
plt.xlabel('Drug-Like')
plt.ylabel('Number of Compounds')
plt.xticks([0,1],['Drug-Like','Not Drug-Like'], rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Molecular_Weight'],bins=10, color='steelblue', edgecolor='black')
plt.axvline(x=500, color='red', linestyle= '--', linewidth=2, label='Lipinski Cutoff (500 Da)')
plt.title('Distribution of Molecular Weight in Withanolides')
plt.xlabel('Molecular Weight (Da)')
plt.ylabel('Number of Compunds')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.patches as mpatches

plt.figure(figsize=(8,5))
colors=df['Drug_Like'].map({True:'steelblue', False: 'salmon'})
plt.scatter(df['Molecular_Weight'], df['XLogP'], c=colors, edgecolor='black', s=100, alpha=0.5)
plt.axvline(x=500, color='red', linestyle ='--', linewidth=1.5)
plt.axhline(y=5, color='orange', linestyle = '--', linewidth=1.5)
plt.title('Molecular Weight vs XLogP coloured by Drug-Likeness')
plt.xlabel('Molecular Weight (Da)') 
plt.ylabel('XLogP')

drug_like_patch = mpatches.Patch(color='steelblue', label='Drug-Like')
not_drug_like_patch = mpatches.Patch(color='salmon', label='Not Drug-Like')
mw_line = plt.Line2D([0],[0], color='red', linestyle='--', label='MW Cutoff (500)')
logp_line = plt.Line2D([0],[0], color='orange', linestyle='--', label='LogP Cutoff (5)')
plt.legend(handles=[drug_like_patch, not_drug_like_patch, mw_line, logp_line])

plt.tight_layout()
plt.show()
```

```python
drug_like_compounds = df[df['Drug_Like']==True][['Name', 'Molecular_Weight', 'XLogP','H-Bond_Donor_Count', 'H-Bond_Acceptor_Count']]
drug_like_compounds = drug_like_compounds.reset_index(drop=True)
print(f"Drug-Like Withanolides ({len(drug_like_compounds)} compounds):\n")
drug_like_compounds
```

```python
drug_like_compounds.to_csv("drug_like_withanolides.csv", index= False)
print("Results saved!")
```

## Conclusion
-21 withanolide compunds from Ashwagandha were analysed
-Lipinski's Rule of Five was applied to assess drug-likeness
-13 out of 21 (62%) compounds passed all four rules
-Molecular Weight was the only failing criterion - all compounds passed LogP, H-Bond Donor and Acceptor Rules
-Withanolide A, B and D are among the top drug-like candidates
